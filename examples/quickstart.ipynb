# TLabel-Bench Quick Start

This notebook shows how to:
1. Load cross-sensor annotations
2. Compare features across sensors
3. Run material classification
4. Evaluate cross-sensor transfer

```python
import json
from pathlib import Path

# Load annotations for each sensor
annotations_dir = Path('../annotations')
sensors = ['gelsight', 'digit', 'dma', 'xense']

for sensor in sensors:
    ann_path = annotations_dir / sensor / 'annotations.json'
    if ann_path.exists():
        with open(ann_path) as f:
            data = json.load(f)
        n = len(data.get('annotations', []))
        print(f'{sensor}: {n} annotations')
```

```python
# Compare features across sensors for the same object
import pandas as pd

rows = []
for sensor in sensors:
    ann_path = annotations_dir / sensor / 'annotations.json'
    if not ann_path.exists():
        continue
    with open(ann_path) as f:
        data = json.load(f)
    for ann in data.get('annotations', []):
        feat = ann.get('features', {})
        rows.append({
            'sensor': sensor,
            'object_id': ann['object_id'],
            'material': ann['material_label'],
            **feat
        })

df = pd.DataFrame(rows)
df.head(12)
```

```python
# Cross-sensor feature comparison for obj_001 (rigid aluminum)
obj_001 = df[df['object_id'] == 'obj_001']
print('Feature comparison for aluminum block across sensors:')
print(obj_001[['sensor', 'eccentricity', 'skewness', 'contact_ratio', 'slip_entropy']].to_string(index=False))
```

## Key Insight

Notice how the same object (aluminum block) produces similar feature patterns across different sensors. This is the core value of TLabel's sensor-agnostic approach — the same 18-dimensional feature vector works regardless of which sensor captured the data.
