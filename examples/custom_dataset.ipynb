```python
# Step 4: Format for TLabel-Bench
import json

def create_benchmark_annotation(object_id, material_label, material_detail,
                                interaction_id, episode, quality_score, features):
    """Create a TLabel-Bench compatible annotation entry."""
    quality_level = (
        'excellent' if quality_score >= 0.8 else
        'good' if quality_score >= 0.6 else
        'acceptable' if quality_score >= 0.4 else
        'poor'
    )
    return {
        'object_id': object_id,
        'material_label': material_label,
        'material_detail': material_detail,
        'interaction_id': interaction_id,
        'episode': episode,
        'quality_score': quality_score,
        'quality_level': quality_level,
        'features': features,
    }

print('Annotation formatter ready. See docs/annotation_schema.md for details.')
```
